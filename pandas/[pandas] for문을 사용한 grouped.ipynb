# for 문을 사용한 grouped 객체
* 모든 group by 되는 경우에 대해 필터링 역할
* grouped 되는 데이터 프레임에 대해 따로 계산이 필요한 경우 유용.
* ex. 모든 그룹에 대해 전월대비 변화량을 계산 -> 각 group별 기간 계산이 필요.
* 모든 데이터에 대해 필터링 해야 전월 대비 값을 파악할 수 있음.(2025년 1월엔 거래가 있고, 2,3월엔 거래가 없고, 4월에 있는 경우)

```python
import pandas as pd

df = pd.DataFrame({
    "product_category": ["Electronics", "Electronics", "Clothing", "Clothing", "Electronics"],
    "client_name": ["ClientA", "ClientA", "ClientB", "ClientB", "ClientC"],
    "region": ["Seoul", "Busan", "Seoul", "Busan", "Seoul"],
    "transaction_type": ["Sale", "Sale", "Return", "Sale", "Return"],
    "amount": [1000, 1500, 500, 700, 1200]
})
df
```

```python
grouped = df.groupby(["product_category", "client_name"])

for group_values, group_df in grouped:
    print("group_values:", group_values)
    print(group_df, "\n")
```
